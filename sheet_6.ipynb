```python
def naive_monte_carlo(func, random_numbers, a, b):
    current_value = 0.0
    n = len(random_numbers)
    for i in range(n):
        x = a + (b - a) * random_numbers[i]
        current_value += func(x)
    current_value /= n
    return current_value * (b - a)
```

```python

func = lambda x: np.exp(-x**2)
result = naive_monte_carlo(func, random_numbers, -np.pi/2, np.pi/2)
print(result)
```

```python
def variance_naive_monte_carlo(func, random_numbers, a, b, known_mean):
    squared_differences = []
    n = len(random_numbers)
    for i in range(n):
        x = a + (b - a) * random_numbers[i]
        y_i = func(x) * (b - a)
        squared_differences.append((y_i - known_mean) ** 2)
    return(sum(squared_differences) / (n-1))
    
```
